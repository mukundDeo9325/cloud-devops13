## Importing Module 

```python
import numpy as np 
```

```python
a = np.array([3,2,1,2,5,6,7,3,4])
print(a)
print(a.dtype)
print(a.ndim)
print(a.shape)
```

```python
b = np.array([
    [1.1,2,3],
    [2,3,4],
    [4,5,6]
]
)
```

```python
print(b)
print(b.dtype)
print(b.ndim)
print(b.shape)
```
