```python
import subprocess as pros
import shutil 

managers = {
    "apt-get": [["apt-get", "update"], ["apt-get","install","-y","nginx"]],
    "yum": [["yum", "update",], ["yum", "install", "-y","nginx"]],
    "dnf": [["dnf", "install", "-y", "nginx"]]
}


for name, commands in managers.items():
    if shutil.which(name):
        for cmd in commands:
            pros.run(cmd, check=True)
    else:
        print("pakage installation compled")    
```
