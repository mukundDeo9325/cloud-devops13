## How to Manage Errors in Python 


```python
## Doutfull code ---> try 
## Code correct ---> expected output 
## code Errors ---> the is buggee 
```

```python
try : 
    z = int(input("enter a value: "))
    print(z)

except:
    print("this is not an valid value ")
```

```python
try : 
    if a > 5:
     print("hello")

except:
    print("this is incorrect")
```

```python
try : 
    z = int(input('Enter an Value: '))
    print(10/z)
except: 
    print("we cannot divide any value by 0")
    
```
